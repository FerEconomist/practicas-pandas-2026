## 1. Obtener datos "sin internet": Lectura de archivos locales

El punto de partida más común en el procesamiento de datos es trabajar con un archivo que ya descargamos y tenemos guardado en nuestra propia computadora. El formato rey en el mundo de los datos es el **CSV** (Valores Separados por Comas), porque es ligero, universal y cualquier programa lo entiende.

Para leer este archivo, Pandas nos ofrece su herramienta más famosa: `pd.read_csv()`. 
Lo que hace esta función es tomar ese archivo de texto plano aburrido y transformarlo instantáneamente en un **DataFrame** (nuestra tabla interactiva de Pandas con filas, columnas e índices).

*Nota: En el siguiente bloque de código, primero vamos a crear un archivo CSV de prueba en la computadora para luego simular cómo se lee en un caso real.*

```python
import pandas as pd

# --- PARTE A: Creamos un archivo CSV falso para hacer la prueba ---
# (En la vida real, este paso no lo haces porque el archivo ya te lo envían armado)
datos_falsos = pd.DataFrame({
    'id_estudiante': [1, 2, 3],
    'materia': ['Matemática', 'Historia', 'Programación'],
    'calificacion': [8.5, 9.0, 10.0]
})

# Guardamos este dataframe como un archivo físico en la carpeta
datos_falsos.to_csv('datos_locales_prueba.csv', index=False)


# --- PARTE B: Lo que realmente importa: Leer el archivo ---
# Así es como traes un archivo de tu disco duro al entorno de Python
ruta_del_archivo = 'datos_locales_prueba.csv'

# Pandas lee el archivo y lo convierte en un DataFrame
df_local = pd.read_csv(ruta_del_archivo)

print("Datos cargados desde el archivo local en tu computadora:")
display(df_local)
```

## 2. Orígenes de datos: ¿Librerías vs. URLs?

Cuando no tenemos el archivo guardado localmente, solemos recurrir a dos fuentes principales. Aunque parezcan similares porque ambas las traemos con código, funcionan de manera muy distinta:

### A. Datasets integrados en Librerías (Ej: Seaborn, Scikit-Learn)
* **¿Qué son?** Son tablas clásicas (como el famoso dataset del Titanic o el de las flores Iris) que ya vienen metidas adentro de la instalación de ciertas librerías.
* **Características:** 
  * **¡No necesitan internet!** Como ya están en tu disco duro desde que instalaste la librería, puedes cargarlos estando desconectado.
  * Son datos "de laboratorio". Vienen limpios y preparados exclusivamente para que practiques.

### B. Repositorios mediante URL (Ej: GitHub, APIs del Gobierno)
* **¿Qué son?** Son archivos que viven en un servidor externo. En lugar de pasarle a `read_csv` la ruta de tu computadora, le pasas un link de internet.
* **Características:**
  * **Dependencia de internet:** Si no hay conexión o el servidor se cae, tu código va a dar error y no podrás cargar la tabla.
  * Es el mundo real. Te conectas a datos vivos que pueden actualizarse constantemente.

```python
# --- EJEMPLO A: Dataset desde una Librería (Sin internet necesario al ejecutar) ---
# Usamos seaborn solo para traer uno de sus datasets de práctica
!pip install seaborn
import seaborn as sns

# Cargamos el famoso dataset de los pinguinos. 
df_libreria = sns.load_dataset('penguins')

print("1. Dataset cargado desde la librería Seaborn (Primeras 3 filas):")
display(df_libreria.head(3))


# --- EJEMPLO B: Dataset mediante una URL (Requiere internet activo) ---
# Usamos un link que apunta directamente a un archivo CSV alojado en internet (GitHub)
url_repositorio = 'https://raw.githubusercontent.com/pandas-dev/pandas/main/doc/data/titanic.csv'

# read_csv funciona exactamente igual para links de internet
df_url = pd.read_csv(url_repositorio)

print("\n2. Dataset cargado en vivo desde internet mediante URL (Primeras 3 filas):")
display(df_url.head(3))
```

## 3. El control de los datos: `index` y `dtype`

A veces, si dejamos que Pandas haga todo en piloto automático, toma decisiones que pueden arruinar nuestros datos o ensuciarlos. Para evitarlo, usamos parámetros específicos al guardar y al leer.

### A. El misterio del índice (`index=False`)
Cuando usamos `to_csv()` para guardar una tabla, Pandas por defecto incluye los números de fila (0, 1, 2...) como si fueran una columna más. Si volvemos a leer ese archivo, nos aparecerá una columna basura llamada "Unnamed: 0". 
* **Solución:** Agregar `index=False` al guardar evita que se exporte esa numeración inútil.

### B. Evitar que Pandas sea "adivino" (`dtype={'id': 'string'}`)
Cuando usamos `read_csv()`, Pandas trata de adivinar el tipo de dato. Si ve una columna de IDs (ej: `0045`), va a decir: *"¡Ah, esto es un número entero!"*.
* **El peligro:** Al leerlo como número, Pandas borra los ceros a la izquierda y lo convierte en `45`. ¡Acabas de perder la trazabilidad de tu ID original! Además, los IDs son etiquetas, no números matemáticos (no vas a sumar dos IDs).
* **Solución:** Le pasamos `dtype={'id': 'string'}`. Le damos la orden estricta de no adivinar y tratar esa columna como texto puro.

```python
# 1. Creamos un archivo de prueba donde un ID tiene ceros a la izquierda
# Lo escribimos manualmente para simular un CSV problemático
with open('datos_con_ceros.csv', 'w', encoding='utf-8') as f:
    f.write('id,nombre\n0015,Ana\n0082,Juan\n0103,María')

print("--- LECTURA AUTOMÁTICA (El error de dejar que Pandas adivine) ---")
# Pandas asume que 'id' es numérico y le borra los ceros
df_malo = pd.read_csv('datos_con_ceros.csv')
display(df_malo)
print("Tipo de dato que asignó a id:", df_malo['id'].dtype) # int64 (entero)


print("\n\n--- LECTURA CONTROLADA (Usando dtype) ---")
# Le damos la orden estricta de que 'id' sea tratado como texto (string)
df_bueno = pd.read_csv('datos_con_ceros.csv', dtype={'id': 'string'})
display(df_bueno)
print("Tipo de dato que asignó a id:", df_bueno['id'].dtype) # string (texto)
```

## 4. Obtención vs. Modificación: La regla de la doble copia

En la teoría de datos, hay una barrera estricta entre los datos crudos (lo que obtenemos) y los datos procesados (lo que modificamos). 

¿Por qué el apunte hace dos copias y le agrega una columna a la segunda? **Lo hace simplemente como una demostración de seguridad.** 
La regla de oro es: apenas leemos un archivo, lo guardamos en una variable (ej: `df_original`) y la dejamos intacta como referencia. Inmediatamente hacemos una copia (`df_trabajo = df_original.copy()`) y trabajamos **solo** sobre esa copia. 

Si nos equivocamos filtrando o borrando cosas, siempre podemos volver a la referencia sin tener que volver a conectarnos a internet o leer el disco duro. Agregarle una nueva columna a la copia de trabajo es la forma visual de comprobar que nuestro `df_original` sigue a salvo.

```python
# Simulamos nuestro archivo recién leído
df_original = pd.DataFrame({'id': ['001', '002'], 'valor': [100, 200]})

# 1. Hacemos la copia de trabajo
df_trabajo = df_original.copy()

# 2. Modificamos SOLO la de trabajo (agregando una columna calculada)
df_trabajo['valor_con_impuesto'] = df_trabajo['valor'] * 1.21

print("--- DATOS ORIGINALES (Intactos, nuestra red de seguridad) ---")
display(df_original)

print("\n--- DATOS DE TRABAJO (Modificados) ---")
display(df_trabajo)
```

## 5. Trazabilidad manual: La Bitácora y el Resumen de Carga

Cuando trabajas en grandes empresas, existen programas automáticos que registran cada clic y cambio que le haces a un dato (trazabilidad). Pero cuando trabajamos localmente, construir una **bitácora manual** (una simple tabla) es una excelente práctica. Nos permite documentar qué decisiones tomamos y, lo más importante, *por qué* las tomamos.

Además, apenas cargamos los datos y antes de ponernos a analizar a fondo, hacemos un **Resumen de Carga**. Es un control de calidad rápido (usando la función `len()` que cuenta la cantidad de filas) para saber si trajimos todo lo que debíamos traer o si hay registros duplicados.

```python
# --- 1. LA BITÁCORA DE TRAZABILIDAD ---
# Creamos un registro de nuestras decisiones importantes
pasos = {
    'Paso': [1, 2],
    'Acción': ['Leer ID como texto', 'Crear copia de trabajo'],
    'Motivo': ['Conservar los ceros a la izquierda', 'Preservar el origen intacto']
}
bitacora_df = pd.DataFrame(pasos)

print("📖 Bitácora del Proyecto:")
display(bitacora_df)


# --- 2. RESUMEN DE CARGA (Control de calidad rápido) ---
# Usamos un dataset de prueba con una fila repetida a propósito
df_entrada = pd.DataFrame({'id': ['001', '002', '003', '003']})

total_filas = len(df_entrada)
filas_unicas = len(df_entrada['id'].unique()) # Cuenta cuántos IDs distintos hay

print("\n📊 Resumen de Carga:")
print(f"- Filas recibidas en total: {total_filas}")
print(f"- Registros (IDs) únicos: {filas_unicas}")

if total_filas != filas_unicas:
    print("⚠️ ALERTA: La cantidad de filas no coincide con los IDs únicos. Hay duplicados.")
```

## 6. El ciclo iterativo: Explorar obliga a retroceder

Finalmente, un principio fundamental del procesamiento de datos: **esto no es una línea recta**. 

A veces avanzamos hasta el paso de exploración o análisis y descubrimos un error estructural. Por ejemplo:
* Descubrimos que toda una columna dice "NaN" porque el archivo de otro sistema usaba punto y coma (`;`) en lugar de coma (`,`) como separador.
* Vemos que las fechas no tienen sentido porque el archivo original estaba en formato estadounidense (Mes/Día/Año).

Cuando pasa esto, **no lo arreglamos con parches al final del código**. La buena práctica nos dice que debemos "retroceder": volvemos a la celda de carga (`pd.read_csv()`) y ajustamos los parámetros desde la raíz (ej: agregando `sep=';'` o configurando el formato de fecha).
