Series (1D): Es una estructura unidimensional similar a una columna o una lista homogénea de datos, donde cada elemento tiene una etiqueta llamada índice (que puede ser numérico, de texto o fechas).

DataFrame (2D): Más que una «serie bidimensional», formalmente es una tabla bidimensional compuesta por una colección o diccionario de Series que comparten el mismo índice de filas.

A. Automático por defecto (RangeIndex)
Si no especificás ningún índice al crear una Series o un DataFrame, Pandas asigna automáticamente un índice entero consecutivo que empieza en 0 (RangeIndex(0, 1, 2, ...)):

```python
# Series con índice automático: [0, 1, 2]
import pandas as pd
s = pd.Series([10, 20, 30])

# DataFrame con índice automático: [0, 1]
df = pd.DataFrame({'nombre': ['Ana', 'Luis'], 'edad': [25, 30]})
print(s)
print (df)
```

B. Al momento de la creación (con el argumento index)
Podés pasarle una lista, tupla o array con las etiquetas deseadas:

```python
# En una Serie
s1 = pd.Series([10, 20, 30], index=['a', 'b', 'c'])

# En un DataFrame
df1 = pd.DataFrame(
    {'edad': [25, 30], 'ciudad': ['Córdoba', 'Rosario']},
    index=['usr_1', 'usr_2']
)
print(s1)
print (df1)
```

2. ¿Cómo se puede cambiar o manipular el índice? SOLO EXPLICACION en texto

Opción A (Convertir una columna en índice): Tomás una columna que ya contiene tus datos (como un DNI, un código o una fecha) y la promovés para que pase a ser el nombre o identificador oficial de cada fila.

Opción B (Resetear o reiniciar el índice): Deshacés el índice actual para volver a la numeración tradicional que empieza en cero (0, 1, 2...). El índice que tenías antes no se pierde: vuelve a convertirse en una columna común de la tabla (a menos que decidas borrarlo).

Opción C (Reemplazar todo el índice a mano): Le asignás directamente una lista completa con nuevos nombres a las filas. Para que funcione, la cantidad de nombres nuevos que entregues debe coincidir exactamente con la cantidad de filas que tiene la tabla.

Opción D (Renombrar filas puntuales): Cambiás el nombre únicamente de filas específicas que elijas (por ejemplo, corregir un error en una fila particular o actualizar una etiqueta), dejando intactas todas las demás.

Opción E (Reindexar): Reorganizás el orden de las filas según una lista de nombres que vos decidas. Si incluís un nombre de fila que antes no existía en los datos, se crea una fila nueva rellenada con espacios vacíos.
