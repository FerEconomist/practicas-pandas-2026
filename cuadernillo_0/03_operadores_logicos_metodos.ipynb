DATAFRAME para ejemplo

```python
import pandas as pd
import numpy as np  # Lo importamos para poder crear un dato nulo (NaN)

datos_portafolio = {
    'Ticker': ['AAPL', 'TSLA', 'GGAL', 'YPFD', 'KO', 'MELI'],
    'Tipo': ['CEDEAR', 'CEDEAR', 'Accion Local', 'Accion Local', 'CEDEAR', 'CEDEAR'],
    'Sector': ['Tecnologia', 'Autos', 'Finanzas', 'Energia', 'Consumo', 'Tecnologia'],
    'Precio': [150.5, 310.0, 3100.0, 15500.0, 60.2, np.nan], # MELI no tiene precio cargado
    'Volumen_Diario': [5000, 3200, 1500, 8000, 4100, 1200],
    'En_Cartera': [True, False, True, True, False, False]
}

df_activos = pd.DataFrame(datos_portafolio)
print(df_activos)
```

A. Operadores de Comparación (Relacionales)
Comparan el valor de una columna contra otro valor (o contra otra columna) y responden "Verdadero" o "Falso".

    == (Exactamente igual a)

    != (Distinto de)

    > y >= (Mayor que / Mayor o igual que)

    < y <= (Menor que / Menor o igual que)


<hr style="border: 1px solid #3498db;">

1. Igualdad (==)
Sirve para buscar coincidencias exactas. Es muy sensible a las mayúsculas y espacios: si buscás 'Cedear' pero el dato dice 'CEDEAR', no lo va a encontrar.

```python
# Queremos ver únicamente los CEDEARs
filtro_cedear = df_activos['Tipo'] == 'CEDEAR'

# Le pasamos el filtro al DataFrame para ver el resultado real
print(df_activos[filtro_cedear])
print('----------------------------')
print()

# Forma directa (la más usada en la práctica, escribiendo todo junto):
print(df_activos[df_activos['Tipo'] == 'CEDEAR'])
```

2. Desigualdad (!=)
Es la otra cara de la moneda. Lo usamos cuando queremos excluir algo específico y dejar todo lo demás.

```python
# Traer todos los activos EXCEPTO los del sector Tecnología
print(df_activos[df_activos['Sector'] != 'Tecnologia'])
```

3. Mayor y Menor (>, <, >=, <=)
Se usan habitualmente con columnas numéricas, como precios, porcentajes o volúmenes.

Python

```python
# Activos cuyo precio de cotización supere los 1000 pesos
print(df_activos[df_activos['Precio'] > 1000])
print('----------------------------')
print()

# Activos con un volumen diario de operaciones menor o igual a 3200
print(df_activos[df_activos['Volumen_Diario'] <= 3200])
```

Un detalle sobre las columnas booleanas (True/False)
En nuestro DataFrame tenemos la columna En_Cartera, que ya contiene valores True y False. Para filtrar por esta columna, no hace falta usar == True, porque Pandas ya entiende el valor lógico. Podés pasarle la columna directamente:

```python
# Traer solo los activos que efectivamente tenemos comprados (True)
print(df_activos[df_activos['En_Cartera']])
```

<hr style="border: 1px solid #3498db;">

1. Operador & (AND / Y)
Se usa cuando necesitás que se cumplan todas las condiciones estrictamente al mismo tiempo.

```python
# Queremos los activos que sean CEDEARs Y que además cuesten menos de 200
filtro_and = (df_activos['Tipo'] == 'CEDEAR') & (df_activos['Precio'] < 200)

print(df_activos[filtro_and])
```

2. Operador | (OR / O)
Se usa cuando alcanza con que se cumpla al menos una de las condiciones. Se hace con la barra vertical (en la mayoría de los teclados está al lado del 1 o con Alt Gr).

```python
# Queremos activos que sean del sector Energía O que tengan un volumen mayor a 4000
filtro_or = (df_activos['Sector'] == 'Energia') | (df_activos['Volumen_Diario'] > 4000)

print(df_activos[filtro_or])
```

3. Operador ~ (NOT / Negación)
Sirve para invertir una condición o una máscara booleana. Lo que era verdadero pasa a ser falso. Es la tecla de la "virgulilla" (la rayita de la ñ).

```python
# Supongamos que queremos ver los activos que NO tenemos en cartera.
# En vez de escribir (df_activos['En_Cartera'] == False), podemos negar la columna directamente:
print(df_activos[~df_activos['En_Cartera']])

# También sirve para negar condiciones complejas enteras:
# "Traeme todo lo que NO sea un CEDEAR"
print(df_activos[~(df_activos['Tipo'] == 'CEDEAR')])
```

1. Filtrar por múltiples valores a la vez: .isin()
El problema: Si querés buscar activos que pertenezcan al sector "Tecnologia", al sector "Energia" o al sector "Consumo", tendrías que escribir una cadena larguísima de condiciones con el operador OR (|):
(df['Sector'] == 'Tecnologia') | (df['Sector'] == 'Energia') | (df['Sector'] == 'Consumo')

La solución: .isin() (del inglés is in, "está en") toma una lista normal de Python con todas las opciones que te sirven y busca si el valor de cada celda coincide con alguno de los elementos de esa lista.

Detalle técnico: .isin() es perfecto para trabajar con variables categóricas (textos que se repiten) y hace que el código sea mucho más limpio y fácil de modificar en el futuro.

```python
# 1. Definimos las opciones que nos interesan
sectores_buscados = ['Tecnologia', 'Energia', 'Consumo']

# 2. Aplicamos el filtro
print(df_activos[df_activos['Sector'].isin(sectores_buscados)])
```

2. Manejo de datos faltantes: .isna() y .notna()
El problema: En nuestro DataFrame, el activo MELI no tiene precio (es un valor nulo, representado matemáticamente como NaN o Not a Number). En Python, un NaN nunca es igual a otro NaN. Si intentás buscar ese error haciendo df_activos['Precio'] == np.nan, Pandas te va a devolver falso en todas las filas. Los nulos rompen la lógica tradicional.

La solución: Se diseñaron métodos que saben detectar la ausencia de datos sin usar matemática.

.isna() (is Not Available): Te devuelve Verdadero solo en las filas donde falta el dato. Es tu herramienta de diagnóstico para encontrar qué registros están incompletos.

```python
# Encontramos exactamente qué activos tienen problemas de carga en el precio
print(df_activos[df_activos['Precio'].isna()])
```

.notna() (is NOT Not Available): Hace lo contrario. Te devuelve Verdadero en las filas donde el dato sí existe y es válido. Es tu herramienta de limpieza rápida para descartar la basura y quedarte con los registros útiles.

```python
# Descartamos los activos sin precio para poder hacer cálculos matemáticos tranquilos
df_limpio = df_activos[df_activos['Precio'].notna()]
print(df_limpio)
```

3. Filtrar intervalos numéricos: .between(a, b)
El problema: Para buscar un precio que esté entre $100 y $500, tendrías que repetir la referencia a la columna usando el operador AND (&):
(df['Precio'] >= 100) & (df['Precio'] <= 500)

La solución: .between() simplifica la búsqueda de rangos numéricos (o de fechas). Toma dos valores límite (el inicio y el fin) y devuelve las filas que caen en el medio.

```python
# Activos cuyo precio se encuentra entre 100 y 400
print(df_activos[df_activos['Precio'].between(100, 400)])
```

Detalle técnico: Por defecto, .between() incluye los extremos (es decir, funciona como >= y <=). Si quisieras que sea estricto (que no incluya exactamente el 100 ni el 400), podés agregarle el parámetro inclusive='neither'.
