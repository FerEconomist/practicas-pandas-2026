EMPIEZO CON SERIES 

1. Creación de las dos Series


```python
import pandas as pd

# SERIE 1: Índice natural (0, 1, 2, 3...)
rendimientos = pd.Series([1.2, -0.5, 2.1, 0.8, -1.1, 3.4, 0.2, 1.5, -0.3, 2.7])
print("--- Serie 1: Índice Natural ---")
print(rendimientos)

# SERIE 2: Índice de etiquetas (fechas en formato string)
volumen_operaciones = pd.Series(
    [1500, 2300, 1800, 4200, 3100, 1900, 2750, 3300, 2100, 4500], 
    index=['01-Sep', '02-Sep', '03-Sep', '04-Sep', '05-Sep', 
           '08-Sep', '09-Sep', '10-Sep', '11-Sep', '12-Sep']
)
print("\n--- Serie 2: Índice de Etiquetas ---")
print(volumen_operaciones)
```

2. Acceso con corchetes [] en ambas Series
Como bien dedujiste, la lógica de diccionario se aplica a ambas, la única diferencia es qué le pasás adentro de los corchetes: el número de posición (Serie 1) o la etiqueta de texto (Serie 2).

A. Accediendo a la Serie 1 (Índice Natural)
Aquí, el índice es numérico, así que usamos el número exacto (empezando por cero):

```python
# Traer el primer elemento
print(rendimientos[0])  # Salida: 1.2

# Traer el quinto elemento (índice 4)
print(rendimientos[4])  # Salida: -1.1

# Modificar el tercer elemento
rendimientos[2] = 5.0
print(rendimientos[2])  # Salida: 5.0
```

B. Accediendo a la Serie 2 (Índice de Etiquetas)
Acá usamos el nombre de la etiqueta que definimos, exactamente como si fuera la clave de un diccionario de Python:

```python
# Traer el volumen operado el 4 de septiembre
print(volumen_operaciones['04-Sep'])  # Salida: 4200

# Modificar un valor usando la etiqueta
volumen_operaciones['08-Sep'] = 2000
print(volumen_operaciones['08-Sep'])  # Salida: 2000

# Agregar una fecha nueva que no existía
volumen_operaciones['15-Sep'] = 3800
print(volumen_operaciones['15-Sep'])  # Salida: 3800
```

SLICING

¿Por qué necesitamos loc e iloc entonces?
Si los corchetes simples [] funcionan tan bien, ¿para qué existen loc e iloc?

El problema surge con el Slicing (intervalos) y la ambigüedad. Imaginate que a la Serie 2, en lugar de fechas ('01-Sep'), le hubieras puesto como etiquetas números desordenados (ej: años 2024, 2021, 2025).

Si vos le decís a Pandas mi_serie[1], Pandas se confunde: ¿Le estás pidiendo el elemento que está en la posición 1 (el segundo elemento) o le estás pidiendo el elemento cuya etiqueta es el número 1?

Para resolver esa confusión de manera elegante y permitir cortes complejos, Pandas creó dos herramientas especializadas:

       .loc (Location): Busca estrictamente por las etiquetas que vos inventaste.

       .iloc (Integer Location): Busca estrictamente por la posición matemática (0, 1, 2...), ignorando por completo qué etiqueta tenga la fila.

El concepto principal que tenés que recordar acá es la diferencia de enfoque y un detalle clave sobre dónde terminan:

     .iloc (Integer Location): Se maneja estrictamente por las posiciones numéricas ocultas de la Serie (0, 1, 2...). Funciona exactamente igual que las listas clásicas de Python: el último número del intervalo NO se incluye.

     .loc (Location): Se maneja estrictamente por las etiquetas que vos le pusiste. A diferencia de Python clásico, en .loc el último elemento del intervalo SÍ se incluye.

```python
# SERIE 1: Queremos los primeros 3 rendimientos (posiciones 0, 1 y 2). 
# Ponemos 0:3 porque el 3 NO se incluye.
print("--- Primeros 3 rendimientos con iloc ---")
print(rendimientos.iloc[0:3])

# SERIE 2: Queremos las posiciones de la 2 a la 4 (es decir, el 3er, 4to y 5to día).
# Ponemos 2:5 porque el 5 NO se incluye.
print("\n--- Días en las posiciones 2, 3 y 4 con iloc ---")
print(volumen_operaciones.iloc[2:5])
```

```python
# SERIE 2 (Índice de fechas): Queremos ver el volumen operado desde el 3 de septiembre hasta el 8 de septiembre.
# Con .loc, la etiqueta final SÍ se incluye.
print("--- Volumen del 03-Sep al 08-Sep con loc ---")
print(volumen_operaciones.loc['03-Sep':'08-Sep'])
```

¿Qué pasa si usamos .loc en la Serie 1 (que tiene índice natural)?
Como las etiquetas de la Serie 1 son los mismos números naturales (0, 1, 2...), .loc los va a buscar por su nombre (etiqueta '0', etiqueta 

```python
# SERIE 1 (Índice natural): Queremos las etiquetas desde la 0 hasta la 2.
# OJO: Como usamos .loc, la etiqueta 2 SÍ se incluye en el resultado.
print("\n--- Rendimientos de la etiqueta 0 a la 2 con loc ---")
print(rendimientos.loc[0:2])
```

DATAFRAMES

1. Creación de los dos DataFrames

```python
import pandas as pd

# Datos base: 7 días de cotizaciones para 4 activos
datos_cotizaciones = {
    'Apple': [150, 152, 149, 155, 158, 160, 159],
    'Tesla': [310, 305, 300, 315, 320, 325, 330],
    'Galicia': [3000, 3050, 2980, 3100, 3150, 3200, 3180],
    'YPF': [15000, 15100, 14900, 15300, 15500, 15800, 15700]
}

# Fechas para usar como etiquetas
fechas_semana = ['01-Sep', '02-Sep', '03-Sep', '04-Sep', '05-Sep', '08-Sep', '09-Sep']

# DATAFRAME 1: Índice Natural (Pandas asigna 0, 1, 2, 3, 4, 5, 6)
df_natural = pd.DataFrame(datos_cotizaciones)
print("--- DataFrame 1: Índice Natural ---")
print(df_natural)

# DATAFRAME 2: Índice de Etiquetas (Asignamos las fechas)
df_etiquetas = pd.DataFrame(datos_cotizaciones, index=fechas_semana)
print("\n--- DataFrame 2: Índice de Etiquetas ---")
print(df_etiquetas)
```

2. Uso como diccionario (Selección y modificación de columnas)
En los DataFrames, usar los corchetes simples [] apunta directamente a las columnas. El comportamiento es exactamente idéntico sin importar si el índice es natural o tiene etiquetas. El índice (las filas) viaja pegado a los datos de la columna que extraigas.

A. Extraer una sola columna (Devuelve una Serie)
Pasás el nombre de la columna como si fuera la clave de un diccionario.

```python
# Del DF con índice natural
print(df_natural['Galicia'])

# Del DF con etiquetas (devuelve los mismos valores, pero con las fechas al lado)
print(df_etiquetas['Galicia'])
```

B. Extraer múltiples columnas (Devuelve un DataFrame)
Le pasás una lista de Python dentro de los corchetes (por eso la doble llave [[]]).

```python
# Traer solo las acciones locales
print(df_etiquetas[['Galicia', 'YPF']])
```

C. Crear o modificar columnas
Podés inventar una clave nueva y asignarle datos, o hacer operaciones matemáticas entre columnas existentes.

```python
# Crear una columna con un valor fijo
df_etiquetas['Riesgo_Pais'] = 1500
print("\n--- DataFrame con columna Riesgo_Pais ---")
print(df_etiquetas)

# Crear una columna sumando el valor de los CEDEARs
df_etiquetas['Total_Cedears'] = df_etiquetas['Apple'] + df_etiquetas['Tesla']
print("\n--- DataFrame con columna Total_Cedears ---")
print(df_etiquetas)
```

.iloc (Búsqueda por Posición Matemática)
Solo entiende de números (0, 1, 2...). Ignora por completo cómo se llaman tus filas o columnas. El límite final del intervalo no se incluye.

```python
# 1. Extraer una fila única (ej. el tercer día de cotización, posición 2)
# Funciona igual en ambos DataFrames porque ambos tienen una "posición 2"
print(df_natural.iloc[2])
print()
print(df_etiquetas.iloc[2])
print()

# 2. Slicing de filas (desde la posición 1 hasta la 4, sin incluir la 4)
print(df_etiquetas.iloc[1:4])
print()

# 3. Slicing cruzado (las primeras 3 filas, y las primeras 2 columnas)
print(df_etiquetas.iloc[0:3, 0:2])
```

.loc (Búsqueda por Etiqueta o Nombre)
Busca exactamente el texto o número que figura visualmente en el índice y en las columnas. El límite final del intervalo sí se incluye.

```python
# 1. Extraer una fila única por su nombre
print(df_etiquetas.loc['04-Sep'])
print()

# En el DF natural, la etiqueta SE LLAMA igual que la posición ('3')
print(df_natural.loc[3]) 
print()

# 2. Slicing cruzado en el DF con etiquetas (Desde el 02-Sep al 05-Sep, abarcando de Tesla a YPF)
print(df_etiquetas.loc['02-Sep':'05-Sep', 'Tesla':'YPF'])
print()
# 3. Traer filas salteadas y una sola columna
print(df_etiquetas.loc[['01-Sep', '08-Sep'], 'Apple'])
print()

# 4. Traer todas las filas (:), pero filtrando un rango de columnas
print(df_etiquetas.loc[:, 'Galicia':'YPF'])
```
